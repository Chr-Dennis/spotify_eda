There's a pretty wide range in artist population. It makes sense for more successful artists to have a much higher appearance rate in the data.

## Visualizing the Data

### A histogram regarding the **danceability** of tracks in this data

```python
df['danceability'].hist(bins = 60)
```

### A Box Plot regarding the same **danceability** metric.

```python
sns.boxplot(x = df['danceability'])
```

### Building a Heatmap

```python
df.columns
df_corr = df[['popularity',
       'duration_ms', 'danceability', 'energy', 'key', 'loudness',
       'mode', 'speechiness', 'acousticness', 'instrumentalness', 'liveness',
       'valence', 'tempo', 'time_signature']].corr()
df_corr
```

```python
fig, ax = plt.subplots(figsize = (15,15))
sns.heatmap(df_corr, annot = True, cmap = "mako")
```

It looks like there is a strong negative correlation between **acousticness** and the fields of **energy** and **loudness**, meaning a track with high **acousticness** value would likely have low **energy** and/or **loudness** value, and vice versa. This aligns with initial perception of these fields and answers a previous question regarding their relationships.  

Additionally, **energy** and **loudness** are even more strongly correlated than the two fields are negatively correlated to the **acousticness** field, which aligns with expectations.

### Visualization of **energy** and **acousticness** correlation specifically

```python
sns.pairplot(df[['acousticness', 'energy']], height = 5, aspect = 1.5)
```

```python
df_sample = df[['acousticness', 'energy']].sample(n = 5000)
sns.pairplot(df_sample[['acousticness', 'energy']], height = 5, aspect = 1.5, diag_kind = 'kde')
```

### Visualization of **energy** and **acousticness** correlation across 5 genres

```python
df['track_genre'].value_counts()

df_genres = df[df['track_genre'].isin(['pop', 'folk', 'emo', 'punk', 'electronic'])]

df_sample = df_genres.sample(n=5000, random_state=1)

sns.pairplot(df_sample[['acousticness', 'energy', 'track_genre']], height = 5, aspect = 1.5, diag_kind = 'kde', hue='track_genre')
```
